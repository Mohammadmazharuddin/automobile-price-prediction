```python
x = df.drop("price",axis=1)
y = df["price"]
```

```python
print(x.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
x.shape
```

```python
y.shape
```

```python
numeric_columns = x_train.select_dtypes(include=["int64", "float64"]).columns
categorical_columns = x_train.select_dtypes(include=["object"]).columns
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer( transformers = [("num" ,StandardScaler(),numeric_columns),
                                           ("cat",OneHotEncoder(handle_unknown="ignore"),categorical_columns)])
```

```python
preprocessor
```

```python
x_train =preprocessor.fit_transform(x_train)
x_test = preprocessor.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
linear_regression = LinearRegression()
```

```python
linear_regression
```

```python
linear_regression.fit(x_train,y_train)
```

```python
lr_pred = linear_regression.predict(x_test)
lr_pred
```

```python
from sklearn.tree import  DecisionTreeRegressor
regression = DecisionTreeRegressor(random_state= 42)
```

```python
regression
```

```python
regression.fit(x_train,y_train)
```

```python
dt_pred = regression.predict(x_test)
dt_pred
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
random = RandomForestRegressor(n_estimators =100,random_state=42)
```

```python
random
```

```python
random.fit(x_train,y_train)
```

```python
rf_pred = random.predict(x_test)
rf_pred
```

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
gr = GradientBoostingRegressor(random_state = 42)
```

```python
gr
```

```python
gr.fit(x_train,y_train)
```

```python
 gr_pred = gr.predict(x_test)
 gr_pred
```

```python
from sklearn.ensemble import ExtraTreesRegressor
```

```python
extra_trees = ExtraTreesRegressor(n_estimators=100,random_state=42)
```

```python
extra_trees 
```

```python
extra_trees
```

```python
extra_trees.fit(x_train,y_train)
```

```python
 et_pred = extra_trees.predict(x_test)
 et_pred
```

```python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
print("Linear Regression :",np.sqrt( mean_squared_error(y_test, lr_pred)))
print("Decision Tree :", np.sqrt(mean_squared_error(y_test, dt_pred)))
print("Random Forest :",np.sqrt(mean_squared_error(y_test, rf_pred)))
print("Gradient Boosting :",np.sqrt(mean_squared_error(y_test, gr_pred)))
print("Extra Trees :",np.sqrt(mean_squared_error(y_test, et_pred)))
```

```python
print("Linear Regression :", r2_score(y_test, lr_pred))
print("Decision Tree :", r2_score(y_test, dt_pred))
print("Random Forest :", r2_score(y_test, rf_pred))
print("Gradient Boosting :", r2_score(y_test, gr_pred))
print("Extra Trees :", r2_score(y_test, et_pred))
```

```python
print("Linear Regression :", mean_absolute_error(y_test, lr_pred))
print("Decision Tree :", mean_absolute_error(y_test, dt_pred))
print("Random Forest :", mean_absolute_error(y_test, rf_pred))
print("Gradient Boosting :", mean_absolute_error(y_test, gr_pred))
print("Extra Trees :", mean_absolute_error(y_test, et_pred))
```

```python
results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest",
        "Gradient Boosting",
        "Extra Trees"
    ],
    
    "MAE": [
        mean_absolute_error(y_test, lr_pred),
        mean_absolute_error(y_test, dt_pred),
        mean_absolute_error(y_test, rf_pred),
        mean_absolute_error(y_test, gr_pred),
        mean_absolute_error(y_test, et_pred)
    ],
    
    "RMSE": [
        np.sqrt(mean_squared_error(y_test, lr_pred)),
        np.sqrt(mean_squared_error(y_test, dt_pred)),
        np.sqrt(mean_squared_error(y_test, rf_pred)),
        np.sqrt(mean_squared_error(y_test, gr_pred)),
        np.sqrt(mean_squared_error(y_test, et_pred))
    ],
    
    "R2": [
        r2_score(y_test, lr_pred),
        r2_score(y_test, dt_pred),
        r2_score(y_test, rf_pred),
        r2_score(y_test, gr_pred),
        r2_score(y_test, et_pred)
    ]
})

results
```
